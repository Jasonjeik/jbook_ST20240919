# Proyecto Final Series de Tiempo

Un caso de estudio de la base de datos de [Kaggle](https://www.kaggle.com/datasets/thomaswrightanderson/river-aire-discharge-time-series?resource=download), para pronósticar el caudal del Air River en Inglaterra. 


## Resumen

Las fuentes hídricas suponen el sustento de la vida en cualquier ecosistema. Estudiar su comportamiento en el tiempo, permite al ser humano tomar decisiones trascendentales que permiten su ploriferación como individuo social. En este proyecto final, se propone utilizar una novedosa forma de Red Neuronal Recurrente (RNN), llamada Red de estado Eco (ESN, por sus siglas en ingles), para la estimación del caudal el Air River en Inglaterra.
Una ESN, a diferencia de las RNN, la mayoría de los pesos de la red están fijos y no se entrenan, lo que reduce significativamente la complejidad del entrenamiento. Sólo los pesos de salida se entrenan, lo cual hace a las ESN, más rápidas y simples en comparación con las RNN tradicionales. Esto para fines prácticos de este proyecto supondrá unas ventajas significativas debido a la limitada capacidad de procesamiento de datos y el tiempo necesario para la ejecución del estudio. 
Las ESN con regulación bayeciana (BESN), ha sido usado en diferentes estudios de sistemas pequenos de plantas hidroeléctricas. En donde, la teoría bayesiana, considera la distribución en el espacio y los pesos de salida se optimizan mediante la maximización de la distribución de probabilidad anterior. 
Se utliza como referencia para la valiación del modelo el caudal del "Air River", medido en dos ciudades Armley y Kildwick, Inglaterra.
